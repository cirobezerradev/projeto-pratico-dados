```python
import duckdb
import pandas as pd
import os
from datetime import datetime
```

## Conexão com o banco de dados DuckDB
Para conectar-se ao banco de dados DuckDB, você pode usar a biblioteca `duckdb` em Python.

```python
conn = duckdb.connect('../databases/data.db')
```

```python
file = 'z0019_2.csv'
data_atual = datetime.now().strftime('%Y-%m-%d %H:%M:%S')
df = pd.read_csv(f'../landing/{file}', sep=';', encoding='utf8')
df['file_name'] = file
df['data_atual'] = data_atual

print(df)
```

```python
conn.execute("""
    CREATE TABLE IF NOT EXISTS bronze_products (
                NATBR VARCHAR,
                MAKTX VARCHAR,
                WERKS VARCHAR,
                MAINS VARCHAR,
                LABST VARCHAR,             
                FILE_NAME VARCHAR,
                DATA_ATUAL TIMESTAMP
    )
""")
```

```python
conn.execute("""INSERT INTO bronze_products SELECT * FROM df""")
```

```python
resultado = conn.execute("""select * from bronze_products""").fetchdf()
print(resultado)
```

```python
conn.execute("""alter table bronze_products rename to bronze_z0019""")
```

```python
conn.close()
```
