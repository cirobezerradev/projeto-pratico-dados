```python
import duckdb
import pandas as pd
```

```python
conn = duckdb.connect('../databases/data.db', read_only=False)
```

```python
df = conn.execute(f"""select * from silver_products""").fetchdf()
df.head(10)
```

```python
df2 = df.drop(columns=['id_category', 'id_supplier'])
df2.head(10)
```

```python
conn.execute("""
CREATE TABLE IF NOT EXISTS dim_products (
    id_product BIGINT,
    nm_product TEXT,
    pr_product FLOAT
)
""")
```

```python
conn.execute("""insert into dim_products select * from df2""")
```

```python
df_dim_products = conn.execute(f"""select * from dim_products""").fetchdf()
df_dim_products.head(10)
```

```python
result = conn.execute("""SELECT table_name, has_primary_key FROM duckdb_tables()""").fetchdf()
print(result)
```
